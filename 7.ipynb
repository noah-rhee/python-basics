**Regular Expressions** are a technique for processing complex strings. They aren't unique to Python — they're used wherever strings need to be processed.

### Why we need it?

Suppose we have the following problem.

> We have some text that contains Korean Resident Registration Numbers (a national ID, formatted like `YYMMDD-XXXXXXX`). Replace the back half (the 7 digits after the dash) of every such number with `*` characters.


Without regular expressions, you'd have to write something like this:

- Split the entire text on whitespace (split).
- For each word, check whether it matches the resident-number format.
- If it does, replace the back half with `*`.
- Reassemble the words.

````python
data = """
park 800905-1049118
kim  700905-1059119
"""

result = []
for line in data.split("\n"):
    word_result = []
    for word in line.split(" "):
        if len(word) == 14 and word[:6].isdigit() and word[7:].isdigit():
            word = word[:6] + "-" + "*******"
        word_result.append(word)
    result.append(" ".join(word_result))
    
print("\n".join(result))
````

```python
Result:
park 800905-*******
kim  700905-*******
```

By contrast, with regular expressions you can write something far simpler and more direct.

We haven't learned the syntax yet, so just take a look for now.

````python
import re 

data = """
park 800905-1049118
kim  700905-1059119
"""

pat = re.compile("(\d{6})[-]\d{7}")
print(pat.sub("\g<1>-*******", data))
````

```python
Result:

```

Regular expressions make even this simple example dramatically shorter. The more complex the pattern you're looking for or replacing, the bigger the payoff.

### Meta Characters

A **meta character** is a character that, instead of meaning the literal character it looks like, is used for a special purpose.

```. ^ $ * + ? { } [ ] \ | ( )```

### Character Class [ ]

The first meta character we'll look at is the character class `[ ]`. A regular expression built from a character class means "match any one of the characters between the `[` and `]`".
<br><br>
> Any character at all can go between the `[` and `]` of a character class.

<br><br>
So the regex `[abc]` means "match exactly one character that is either `a`, `b`, or `c`". To get a feel for it, see how the strings `"a"`, `"before"`, and `"dude"` match against the regex `[abc]`.
<br><br>
- `"a"` matches because it contains `"a"`, which is in the regex.
- `"before"` matches because it contains `"b"`, which is in the regex.
- `"dude"` does not match — it contains none of `a`, `b`, or `c`.<br><br>

If you put a hyphen (`-`) between two characters inside `[ ]`, it indicates a range (From - To). For example, the regex `[a-c]` is the same as `[abc]`, and `[0-5]` is the same as `[012345]`.
<br><br>
Some examples of character classes that use the hyphen:
<br><br>
- `[a-zA-Z]` : all alphabetic characters
- `[0-9]` : digits
<br><br><br><br>

Inside a character class `[ ]` you can use any character — including any meta character — but there's one meta character that needs special attention.

That's the **`^`** character: when used inside a character class, it means **NOT**. For example, the regex `[^0-9]` matches any character that is *not* a digit.

### Dot(.)

The `Dot (.)` meta character in regex matches any character except a newline (`\n`).

> As we'll see later, if you compile the regex with the `re.DOTALL` option, `.` will also match `\n`.

Look at the regex below.

``` a.b ```

This regex means:

> "a + (any character) + b"

In other words, it matches any string with `a` and `b` separated by any single character.

To make this concrete, here's how `"aab"`, `"a0b"`, and `"abc"` line up against the regex `a.b`.

- `"aab"` matches: the middle character `"a"` matches `.`, which means "any character".
- `"a0b"` matches: the middle character `"0"` matches `.`.
- `"abc"` does NOT match: the regex requires exactly one character between `a` and `b`, but here `a` and `b` aren't separated by any character.

Now look at this regex.

```a[.]b```

This regex means:

> ```"a + literal Dot (.) + b"```

So `a[.]b` matches the string `"a.b"` and does NOT match `"a0b"`.

Note that when `.` appears *inside* a character class `[ ]`, it does NOT mean "any character" — it just means a literal dot. Don't get these confused.

### Repetition (*)

Look at this regex.

```ca*t```

This uses the `*` meta character, which means "repetition".

Here `*` means that the character right before it (`a`) can repeat anywhere from 0 times up to (effectively) infinity.

> Technically the upper bound on `*` repetitions isn't truly infinite — due to memory limits, it's around 200 million.

So `cat`, `caat`, `caaaaaaaaaat`, and so on all match.

### Repetition (+)

Another repetition meta character is `+`. `+` means "repeat at least once". In other words, where `*` allows 0 or more repetitions, `+` requires 1 or more.

Look at this regex.

```ca+t```

This means:

> "c + a (1 or more times) + t"

So `cat` and `caaaat` match, but `ct` does NOT match because `a` appears 0 times.

### Repetition ({m,n}, ?)

Now think about this for a moment. What if you want to limit the repetition to exactly 3 times, or to between 1 and 3 times?

The `{ }` meta characters let you fix the repetition count. The regex `{m, n}` matches between m and n repetitions. You can also leave out either `m` or `n`. Writing `{3,}` means 3 or more, and writing `{,3}` means 3 or fewer. Omitted `m` is treated as 0, and omitted `n` is treated as infinity (well, less than 200 million).

> `{1,}` is equivalent to `+`, and `{0,}` is equivalent to `*`.


<br><br>

Let's look at a few regexes that use `{ }`.

**1. {m}**

```ca{2}t```

This regex means:

> "c + a (exactly 2 times) + t"

So `cat` does NOT match (only one `a`), but `caat` does match (two `a`s).

**2. {m, n}**

```ca{2,5}t```

This regex means:

> "c + a (2 to 5 times) + t"

So `cat` does NOT match, but `caat` and `caaaaat` do.

**3. ?**

Although it isn't strictly repetition, `?` is a similar concept. The `?` meta character means ```{0, 1}```.

Look at the regex below.




``` ab?c```

This regex means:

> "a + b (which may or may not appear) + c"

So `abc` matches (`b` appears once), and `ac` also matches (`b` appears 0 times).

### Compile options

When compiling a regular expression you can use the following options:

- `DOTALL` (`S`) — make `.` match any character, including newlines.
- `IGNORECASE` (`I`) — match without regard to case.
- `MULTILINE` (`M`) — allow matching across multiple lines (relates to how `^` and `$` work).
- `VERBOSE` (`X`) — turn on verbose mode (lets you write more readable regexes with whitespace and comments).

When using these options you can write the full name like `re.DOTALL` or the abbreviation like `re.S`.

**1. DOTALL, S**

By default, the `.` meta character matches any character *except* a newline (`\n`). If you want it to match `\n` too, compile the regex with the `re.DOTALL` (or `re.S`) option.

See the example below.

````python
import re
p = re.compile('a.b')
m = p.match('a\nb')

print(m)
````

With the regex `a.b`, the string `a\nb` does NOT match — because `\n` doesn't match the `.` meta character. To make it match `\n` too, use the `re.DOTALL` option as shown below.

````python
p = re.compile('a.b', re.DOTALL)
m = p.match('a\nb')

print(m)
````

Typically `re.DOTALL` is used when you want to search across multi-line strings without caring about newlines.

**2. IGNORECASE, I**

`re.IGNORECASE` (`re.I`) makes matches case-insensitive. See the example below.

````python
p = re.compile('[a-z]+', re.I)
p.match('python')

#p.match('Python')
#p.match('PYTHON')
````

The regex `[a-z]+` only covers lowercase letters, but with `re.I` it matches regardless of case.

**3. MULTILINE, M**

`re.MULTILINE` (`re.M`) is related to the meta characters ```^``` and ```$```, which we'll cover shortly. Briefly: ```^``` means "start of the string" and ```$``` means "end of the string".

For example, the regex `^python` requires the string to *start* with `python` to match, and `python$` requires the string to *end* with `python`.

Look at the example below.

````python
import re
p = re.compile("^python\s\w+")

data = """python one
life is too short
python two
you need python
python three"""

print(p.findall(data))
````

The regex `^python\s\w+` means: start with `python`, then a whitespace character, then a word. The string `data` we're searching has multiple lines.

Running this script gives the following result.

````python
['python one']
````

Because of the `^` meta character, only the *first* line — which starts with `python` — matches.

But you might want `^` to mean "start of each line" rather than "start of the entire string". That's exactly what the `re.MULTILINE` (`re.M`) option is for. Let's modify the previous code as follows.

````python
import re
p = re.compile("^python\s\w+", re.MULTILINE)

data = """python one
life is too short
python two
you need python
python three"""

print(p.findall(data))
````

With the `re.MULTILINE` option, `^` now means "start of each line" rather than "start of the whole string". Running this script produces:

````python
['python one', 'python two', 'python three']
````

In other words, `re.MULTILINE` makes `^` and `$` apply to each line of the string.

**4. VERBOSE, X**

The regexes we've looked at so far are pretty simple, but real-world regexes written by experts can look like cryptography. To understand them you have to dissect them piece by piece. Wouldn't it be nice if you could break a hard-to-read regex into commented, line-by-line pieces? You can! Use the `re.VERBOSE` (`re.X`) option.

See the example below.

````python
charref = re.compile(r'&[#](0[0-7]+|[0-9]+|x[0-9a-fA-F]+);')
````

Is that regex easy to understand? Now look at this version.

````python
charref = re.compile(r"""
 &[#]                # Start of a numeric entity reference
 (
     0[0-7]+         # Octal form
   | [0-9]+          # Decimal form
   | x[0-9a-fA-F]+   # Hexadecimal form
 )
 ;                   # Trailing semicolon
""", re.VERBOSE)
````

Compare the first and second examples. The compiled pattern object `charref` does the exact same thing in both cases. But for a complex regex, writing it across multiple lines with comments — like the second version — is much more readable.

When you use the `re.VERBOSE` option, whitespace inside the regex string is ignored at compile time (except whitespace inside `[ ]`), and you can write comments line by line using `#`.

Hint: Use a `while` loop together with the `random` library, ending the loop only once the condition "the two random integers are coprime" is met.

<br><br>
<example>

```
numRange = [2,3,4,5,6]

..some code..


Output:

Trial 1 : 2, 4
Trial 2 : 2, 4
Trial 3 : 3, 6
Trial 4 : 6, 4
Trial 5 : 2, 3
The last try was coprime, so we stop.

```


Note: Since we're only checking two numbers for coprimality, you could just check it with division. But you can also use a built-in function.

However, if you want to use the built-in for greatest common divisor here, you'll also need to use `reduce`. A challenge! If it's too tricky, just use plain division.

Student's solution

````python
import random
import math
numRange = [2,3,4,5,6]
coprime = False
count = 0
while coprime == False:
    count += 1
    a = random.randint(numRange[0], numRange[-1])
    b = random.randint(numRange[0], numRange[-1])
    print(f"Trial {count} : {a}, {b}")
    if math.gcd(a,b) == 1:
        coprime = True
        print("The last try was coprime, so we stop.")
    
        

# Write your code



````
