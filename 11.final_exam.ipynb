**`sorted(iterable, key=...)`** - sorts using a *key function* that tells Python what to sort **by**.

- `key` takes one item and returns the value to compare.
- Return a **tuple** to sort by several things at once: `(first, second)` sorts by `first`, then breaks ties by `second`.
- A leading minus (`-value`) flips one field to descending; add `reverse=True` to flip everything.

````python
words = ["pear", "fig", "apple", "kiwi"]

# sort by length, then alphabetically to break ties
print(sorted(words, key=lambda w: (len(w), w)))
# -> ['fig', 'kiwi', 'pear', 'apple']

# by length DESCENDING, then alphabetically for ties
print(sorted(words, key=lambda w: (-len(w), w)))
# -> ['apple', 'pear', 'kiwi', 'fig']
````

**`filter(function, iterable)`** - keeps only the items for which `function` returns `True`. It returns a *filter object*, so wrap it in `list(...)` (or loop / comprehend over it). The `function` is usually a small `lambda`.

````python
nums = [5, 12, 3, 8, 20]
print(list(filter(lambda x: x >= 8, nums)))   # -> [12, 8, 20]
````

**`collections.Counter`** - counts how often each item appears in an iterable. It behaves like a dictionary `{item: count}`, and `.most_common()` returns `(item, count)` pairs from most to least frequent.

````python
from collections import Counter

c = Counter("banana")
print(c)                 # -> Counter({'a': 3, 'n': 2, 'b': 1})
print(c["a"])            # -> 3
print(c.most_common(2))  # -> [('a', 3), ('n', 2)]
````

**`itertools.product(iterable, repeat=k)`** - every ordered combination *with repetition* - the same thing as `k` nested loops over the iterable.

````python
from itertools import product

for combo in product([0, 1], repeat=2):
    print(combo)
# -> (0, 0) (0, 1) (1, 0) (1, 1)
````

## Problem 1. (15 pts) - Strings, Dictionaries & Loops

Two words are **anagrams** if one is a rearrangement of the other's letters (e.g. `listen` and `silent`).

Write a function `group_anagrams(words)` that groups a list of words into their anagram families.

Rules:

- Grouping is **case-insensitive** (`Silent` and `listen` belong together).
- Each group must be a list of the **original words** (keep their original spelling), sorted **alphabetically**.
- Return a **list of groups**, and sort the groups **alphabetically** (i.e. by their first word).

Example:

```python
print(group_anagrams(["bat", "tab", "cat", "act", "dog"]))
# -> [['act', 'cat'], ['bat', 'tab'], ['dog']]
```

Hint: Two words are anagrams exactly when their **sorted letters** are equal - so `"".join(sorted(word.lower()))` makes a perfect dictionary key. Remember `dict.setdefault(key, [])` or `dict.get`.

Student's solution

````python
# Write your code

def group_anagrams(words):
    d = {}
    
    for i in range(len(words)):
        original = words[i]              
        letters = sorted(original.lower())  
        
        key = ""
        for letter in letters:           
            key += letter
        
        if key in d:
            d[key].append(original)
        else:
            d[key] = [original]
    
    groups = []
    for key in d:
        groups.append(sorted(d[key]))
    
    groups = sorted(groups, key=lambda g: g[0])
    return groups


# TEST CODE #
print(group_anagrams(["bat", "tab", "cat", "act", "dog"]))
# -> [['act', 'cat'], ['bat', 'tab'], ['dog']]


# TEST CODE #
print(group_anagrams(["bat", "tab", "cat", "act", "dog"]))
print(group_anagrams(["listen", "Silent", "hello", "olleh", "world"]))
````

## Problem 2. (15 pts) - Nested Loops & Patterns

Write a function `hollow_diamond(n)` that prints a **hollow diamond** of stars of size `n`.

For `n = 4`, the output must be exactly:

```
   *
  * *
 *   *
*     *
 *   *
  * *
   *
```

Only the **outline** is drawn: every row has a star at each edge and **spaces in between** (the very top and bottom rows are a single star). The figure is symmetric top-to-bottom.

Note: If you do not use loops to generate the stars/spaces (e.g. you hard-code each line), you get **0 points**. The stars and spaces must be generated, not typed out.

Hint: The row 'levels' go `1, 2, ..., n, n-1, ..., 1`. For level `i` the leading spaces are `n - i`; a middle row is `"*" + " " * (2*i - 3) + "*"`, while the single-star tip (`i == 1`) is a special case.

Student's solution

````python
# Write your code

def hollow_diamond(n):
    l = ["   *"]
    num = n-2
    space1 = 1
    for i in range(n-1):
        space = ""
        
        for i in range(num):
            space += " "
        space += "*"
        for i in range(space1):
            space += " "
        space += "*"
        l.append(space)
        space1 += 2
        num -= 1

    for i in l:
        print(i)
    l.reverse()
    for i in range(1,len(l)):
        print(l[i])


# TEST CODE #
hollow_diamond(4)
````

Build a class `Account` that models a simple bank account.

Requirements:

- The **constructor** takes an owner name and an optional starting balance: `Account(owner, balance=0)`.
- `deposit(amount)` adds `amount` to the balance.
- `withdraw(amount)` subtracts `amount` from the balance, but raises a `ValueError` if the amount is **larger than the current balance**.
- Both `deposit` and `withdraw` must raise a `ValueError` if `amount` is **not positive** (`<= 0`).
- `show()` prints the account in the form `owner: balance`.

Example:

```python
a = Account("Zoe", 100)
a.deposit(50)
a.show()          # -> Zoe: 150
a.withdraw(30)
a.show()          # -> Zoe: 120
a.withdraw(1000)  # -> raises ValueError
```

Hint: Store `self.owner` and `self.balance` in the constructor. Every method needs `self` as its first parameter. Use `raise ValueError("...")` to reject bad input.

Student's solution

````python
# Write your code

class Account:
    def __init__(self,owner, balance = 0):
        self.owner = owner
        self.balance = balance
    def deposit(self,amount):
        self.balance += amount
    def withdraw(self,amount):
        if amount > self.balance:
            raise ValueError
        else:
            self.balance -= amount
    def show(self):
        print(f"{self.owner}: {self.balance}")

    pass


# TEST CODE #
a = Account("Zoe", 100)
a.deposit(50)
a.show()
a.withdraw(30)
a.show()
a.withdraw(1000)
````

```python
a = PremiumAccount("Zoe", 1500)

print(a.is_rich())   # -> True

```

Hint: You do **not** need to write a new constructor - `PremiumAccount` reuses `Account`'s. Inside `transfer`, just call `self.withdraw(...)` and `other.deposit(...)`.

Student's solution

````python
# Write your code

class PremiumAccount(Account):
    def transfer(self,other,amount):
        self.withdraw(amount)
        other.deposit(amount)
    def is_rich(self):
        if self.balance > 1000:
            return True
        else:
            return False


# TEST CODE #
a = PremiumAccount("Zoe", 1500)
b = PremiumAccount("Amy", 100)
a.transfer(b, 500)
a.show()
b.show()
print(a.is_rich())
print(b.is_rich())
````

You are given a list of vote strings (candidate names, possibly repeated). Write a function `podium(votes)` that returns the **top 3 names** by number of votes.

- You **must** use `collections.Counter`.
- Sort by **vote count, highest first**; if two names are tied, break the tie by **name, alphabetically**.
- Return a **list** of at most 3 names (fewer if there are fewer than 3 distinct names).

Example:

```python
votes = ["a", "b", "a", "c", "b", "a", "c", "c", "c"]
print(podium(votes))   # -> ['c', 'a', 'b']   (c=4, a=3, b=2)
```

Hint: `Counter(votes).items()` gives `(name, count)` pairs. A `sorted` with `key=lambda kv: (-kv[1], kv[0])` sorts by count descending, then name ascending. Then slice `[:3]`.

Student's solution

````python
from collections import Counter

# Write your code

def podium(votes):
    e = list(Counter(votes).items())
    e = sorted(e, key=lambda kv: (-kv[1], kv[0]))
    f = []
    for i in range(3):
        f.append(e[i])
    g = []
    for i in f:
        g.append(i[0])
    return g
    

# TEST CODE #
print(podium(["a", "b", "a", "c", "b", "a", "c", "c", "c"]))
````

### Problem 4. B. (10 pts) - `itertools.product`

Write a function `dice_ways(n, target)` that returns **how many different ways** you can roll `n` ordinary six-sided dice so that the pips add up to exactly `target`.

- You **must** use `itertools.product` (each die shows `1`-`6`).
- Dice are distinguishable, so `(1, 6)` and `(6, 1)` count as **two** different ways.
- Return an **integer**.

Example:

```python
print(dice_ways(2, 7))   # -> 6   (1+6, 2+5, 3+4, 4+3, 5+2, 6+1)
print(dice_ways(3, 5))   # -> 6
```

Hint: `product(range(1, 7), repeat=n)` yields every possible roll as a tuple. Count the ones whose `sum(...)` equals `target`.

Student's solution

````python
from itertools import product

# Write your code

def dice_ways(n, target):
    t = product(range(1, 7), repeat=n)
    t1 = []
    for i in t:
        count = 0 
        for j in range(n):
            count += i[j]
        if count == target:
            t1.append(i)
    return len(t1)


# TEST CODE #
print(dice_ways(2, 7))
print(dice_ways(3, 5))
````

### Problem 5. B. (10 pts) - Nested comprehension

You are given a **matrix** (a list of lists of integers). Write a function `big_evens(matrix)` that, using a **single nested list comprehension**, returns a **sorted** list containing **every even number** in the matrix, each **doubled**.

You **must** flatten and filter in **one** comprehension (two `for` clauses plus an `if`). No `.append`.

Example:

```python
print(big_evens([[1, 2, 3], [4, 5], [6]]))
# evens are 2, 4, 6 -> doubled -> [4, 8, 12]
```

Hint: A nested comprehension reads left-to-right like nested loops: `[expr for row in matrix for x in row if keep]`.

Student's solution

````python
# Write your code

def big_evens(matrix):
    l =  [ x*2 for row in matrix for x in row if x % 2 == 0       ]
    return sorted(l)


# TEST CODE #
print(big_evens([[1, 2, 3], [4, 5], [6]]))
````

## Problem 6. (10 pts) - Putting It All Together

You are given a list of spending records, each a dictionary:

```python
transactions = [
    {"category": "food", "amount": 10},
    {"category": "toy",  "amount": 30},
    {"category": "food", "amount": 25},
    {"category": "book", "amount": 35},
]
```

Write a function `summary(transactions)` that returns a **list of strings** of the form `"category: total"`, where `total` is the **sum of all amounts** in that category.

- Sort by **total, highest first**.
- If two categories have the **same total**, break the tie by **category name, alphabetically**.

You **must** aggregate the totals with a loop (or built-in), then use `sorted` with a `key` and a **list comprehension** to build the strings.

Example output for the data above:

```python
['book: 35', 'food: 35', 'toy: 30']
```

Hint: Build a `{category: total}` dict first (`totals.get(cat, 0) + amount`). Then `sorted(totals.items(), key=lambda kv: (-kv[1], kv[0]))` gives the right order in one shot.

Student's solution

````python
# Write your code

def summary(transactions):
    d = {}
    for i in transactions:
        if i["category"] in d:
            d[i["category"]] += i["amount"]
        else:
            d[i["category"]] = i["amount"]

    l = []
    e = list(d.keys())
    f = list(d.values())

    for i in range(len(e)):
        tu = (e[i], f[i])
        l.append(tu)
    l = sorted(l, key=lambda kv: (-kv[1], kv[0]))
    l1 = [f"{i[0]}: {str(i[1])}"    for i in l     ]

    return l1


# TEST CODE #
transactions = [
    {"category": "food", "amount": 10},
    {"category": "toy",  "amount": 30},
    {"category": "food", "amount": 25},
    {"category": "book", "amount": 35},
]
print(summary(transactions))
````

## Problem 7. Extra Credit (10 pts) - Perfect Numbers with a List Comprehension

A **perfect number** is a positive integer that equals the sum of all its *proper divisors* (its divisors excluding itself). For example `6 = 1 + 2 + 3`.

Write a function `perfect_numbers(n)` that returns a **list of every perfect number** from `1` up to and including `n`.

- You **must** produce the result with a **single list comprehension**.
- The built-in `sum(...)` over an inner generator of divisors is very handy here.

Example:

```python
print(perfect_numbers(30))   # -> [6, 28]
```

Note: This is **extra credit** - only attempt it after you are confident with Problems 1-6.

Hint: For a number `x`, `sum(d for d in range(1, x) if x % d == 0)` adds up its proper divisors. Drop the check `== x` into the `if` part of a comprehension over `range(1, n + 1)`.

Student's solution

````python
# Write your code

def perfect_numbers(n):
    l = [ i   for i in range(1,n+1)  if sum(d for d in range(1, i) if i % d == 0) == i        ]
    return l


# TEST CODE #
print(perfect_numbers(30))
````
