## 2.Tuple

````python
t1 = ()
t2 = (1, )
t3 = (1, 2, 3)
t4 = 1, 2, 3
t5 = ('a', 'b', ('ab', 'cd'))
````

Tuples look almost the same as lists, but there are 2 differences from lists. When a tuple has only one element, like t2 = (1,), you must add a trailing comma (,) after the element. Also, like t4 = 1, 2, 3, you can omit the parentheses ( ).
<br><br>
At first glance, tuples and lists serve a similar role, but in programming it is better to use tuples and lists differently.

## 3.Dictionary

Anyone can be described with mappings like "name" = "Hong Gildong", "birthday" = "some month and day", and so on. Cleverly, Python has a data type that can express this kind of correspondence. Most modern languages also have such a type for representing correspondences; it is called an associative array or a hash.

<br>

In Python, this type is called a Dictionary. Translated literally, the word means a dictionary (the book). Just like the word "people" maps to "person" and the word "baseball" maps to "the sport baseball", a Dictionary is a data type that holds Key and Value as a pair. For example, if the Key is "baseball", the Value would be "the sport baseball".

<br>

Unlike lists or tuples, dictionaries do not get an element value sequentially; instead, you obtain a Value through a Key. This is the biggest characteristic of a dictionary. To find the meaning of "baseball", you do not scan the entire dictionary sequentially - you flip directly to the page that contains the word "baseball".

````python
dic_1 = {'name':'pey',  'birth': '1118', 'phone':'010-9999-1234'}

dic_2 = {'name':'pey','phone':'010-9999-1234', 'birth': '1118'}


dic_1 == dic_2
````

### Characteristics of a Set

Doesn't the result of ```set("Hello")``` above seem a bit strange?

We clearly built the set type from the string "Hello", but the resulting set is missing one of the l characters and the order is jumbled.

The reason is that a set has the following **two big characteristics**.

- It does not allow duplicates.
- It is unordered.

- Because sets do not allow duplicates, they are often used as a filter to remove duplicates from a data type.

- Like Dictionary, sets do not support indexing because they have no order.

````python
# string

a = 'hello world'
print(a[6]) # indexing
print(a[3:6]) # slicing

# list

b = [a, 156, ['a', 83]]
print(b[1]) # indexing
print(b[1:2]) # slicing
print(b[2][0]) # double indexing

b[1] = 'hello' # modifiable

# tuple

c = (a, 156, ['a', 83])
print(c[1])
print(c[1:2])
print(c[2][0])

c[1] = 'hello' # this can NOT be modified/deleted!

# dictionary

d = {'name': 'junyeong', 'height': '180', 'grade': '4th'}
print(d['name']) # this is not indexing (because there is no notion of order)
print(list(d.keys()))
print(list(d.values()))

'height' in d # whether a certain key exists in the dictionary

# set

e = {} # WRONG. Because {} is the 'operator' for dictionary. The shape is the same, though.
e = set() # this is correct
e = {1, 2, 3} # this works. Because there is no way for Python to get confused.
e = set('jheasdjfjadsl;fjal;kjfk;lsdjfljfal;djf;l')
e = set(['1', 2, 2, 3, 7, '7'])
````
